# Passo 1: Unificação e Limpeza Fina.

```python
print("--- Passo 1: Unificação e Limpeza Fina ---")

import unicodedata

# 1. Função para padronizar nomes de municípios
def normalizar_municipio(valor):
    valor = str(valor).strip().upper()

    valor = unicodedata.normalize(
        'NFKD',
        valor
    )

    valor = ''.join(
        caractere
        for caractere in valor
        if not unicodedata.combining(caractere)
    )

    return valor


# 2. Unir as bases de fraude (2023/2024 + 2025)
df_fraude_completo = pd.concat(
    [
        df_fraude_geral,
        df_fraude_2025
    ],
    ignore_index=True
)


# 3. Corrigir o nome da coluna 'mes ' para 'mes'
df_fraude_completo.rename(
    columns={
        'mes ': 'mes'
    },
    inplace=True
)


# 4. Identificar registros com valores repetidos
duplicados_identificados = (
    df_fraude_completo
    .duplicated()
    .sum()
)

# Os registros repetidos são mantidos porque a base
# não possui identificador único da ocorrência.


# 5. Padronizar os nomes dos municípios
df_fraude_completo['municipio'] = (
    df_fraude_completo['municipio']
    .apply(normalizar_municipio)
)

df_populacao['municipio'] = (
    df_populacao['municipio']
    .apply(normalizar_municipio)
)


# 6. Validar compatibilidade dos municípios
municipios_fraude = set(
    df_fraude_completo['municipio']
    .unique()
)

municipios_populacao = set(
    df_populacao['municipio']
    .unique()
)

municipios_sem_populacao = (
    municipios_fraude
    - municipios_populacao
)

municipios_sem_ocorrencias = (
    municipios_populacao
    - municipios_fraude
)


# 7. Validar os resultados
print(
    f"Dimensão original (soma das bases): "
    f"{len(df_fraude_geral) + len(df_fraude_2025)} linhas"
)

print(
    f"Registros com valores repetidos identificados: "
    f"{duplicados_identificados}"
)

print(
    "Registros repetidos removidos: 0 "
    "(mantidos por ausência de identificador único)"
)

print(
    f"Dimensão final da base unificada: "
    f"{df_fraude_completo.shape}"
)

print(
    f"Anos presentes: "
    f"{sorted(df_fraude_completo['ano'].unique().tolist())}"
)

print(
    f"Municípios na base de ocorrências: "
    f"{len(municipios_fraude)}"
)

print(
    f"Municípios na base populacional: "
    f"{len(municipios_populacao)}"
)

print(
    "Municípios presentes nas ocorrências "
    "e ausentes na população:",
    municipios_sem_populacao
)

print(
    "Municípios presentes na população "
    "e ausentes nas ocorrências:",
    municipios_sem_ocorrencias
)

print(
    f"Colunas atuais: "
    f"{df_fraude_completo.columns.tolist()}"
)

print("\nAmostra dos dados (Head):")
display(
    df_fraude_completo.head(3)
)
```

Nesta etapa, as bases de ocorrências referentes aos anos de 2023, 2024 e 2025 foram unificadas em uma única estrutura.

Durante a preparação dos dados, o nome da coluna do mês foi corrigido de `'mes '` para `'mes'`, evitando inconsistências nas etapas seguintes. Também foram identificados registros com valores repetidos. Como a base disponibilizada não possui um identificador único de ocorrência que permita distinguir com segurança duplicatas reais de ocorrências distintas com atributos coincidentes, esses registros foram preservados.

Além disso, os nomes dos municípios foram normalizados por meio da remoção de espaços nas extremidades, conversão para letras maiúsculas e remoção de acentuação. Essa padronização cria uma chave textual consistente para integração entre a base de ocorrências e a base populacional.


# Passo 2: Construção da Grade Espaço-Temporal e Agrupamento

```python
print("--- Passo 2: Construção da Grade Espaço-Temporal ---")

# 1. Agrupar o total de ocorrências por município, ano e mês
df_ocorrencias_mensais = (
    df_fraude_completo
    .groupby(
        ['ano', 'mes', 'municipio'],
        as_index=False
    )
    .agg(
        total_ocorrencias=('municipio', 'size'),
        ocorrencias_web=('ambiente_web', 'sum')
    )
)

# 2. Criar a lista completa de municípios
municipios = sorted(
    df_populacao['municipio']
    .dropna()
    .unique()
)

# 3. Criar todas as combinações entre:
#    78 municípios x 3 anos x 12 meses
grade_completa = pd.MultiIndex.from_product(
    [
        [2023, 2024, 2025],
        range(1, 13),
        municipios
    ],
    names=[
        'ano',
        'mes',
        'municipio'
    ]
).to_frame(index=False)

# 4. Integrar as ocorrências reais à grade completa
df_agrupado = grade_completa.merge(
    df_ocorrencias_mensais,
    on=[
        'ano',
        'mes',
        'municipio'
    ],
    how='left'
)

# 5. Combinações sem registros de ocorrência
#    representam meses com zero ocorrências
df_agrupado[
    [
        'total_ocorrencias',
        'ocorrencias_web'
    ]
] = (
    df_agrupado[
        [
            'total_ocorrencias',
            'ocorrencias_web'
        ]
    ]
    .fillna(0)
    .astype(int)
)

# 6. Ordenar cronologicamente por município, ano e mês
df_agrupado = (
    df_agrupado
    .sort_values(
        by=['municipio', 'ano', 'mes']
    )
    .reset_index(drop=True)
)

# 7. Validar os resultados
total_grade = (
    df_agrupado['total_ocorrencias']
    .sum()
)

total_base = len(df_fraude_completo)

print(
    f"Total de combinações na grade: "
    f"{len(df_agrupado)} linhas (Municípios x Meses)"
)

print(
    f"Existem valores nulos na contagem? "
    f"{df_agrupado['total_ocorrencias'].isnull().any()}"
)

print(
    f"Total de ocorrências na base unificada: "
    f"{total_base}"
)

print(
    f"Total de ocorrências na grade: "
    f"{total_grade}"
)

print(
    f"Total de crimes agrupados bate com a base original? "
    f"{total_grade == total_base}"
)

print("\nAmostra da Grade Espaço-Temporal (Head):")
display(df_agrupado.head())

print("\nVerificando meses com 'Zero' ocorrências (Exemplo):")
display(
    df_agrupado[
        df_agrupado['total_ocorrencias'] == 0
    ].head(3)
)
```

Nesta etapa, foi construída a grade espaço-temporal contendo todas as combinações entre os 78 municípios e os 36 meses do período analisado, totalizando 2.808 linhas.

As ocorrências foram previamente agrupadas por município, ano e mês e, em seguida, integradas à grade completa. Combinações sem registros foram preenchidas com zero para preservar a continuidade das séries temporais.

A validação compara a soma de `total_ocorrencias` na grade com a quantidade de registros da base unificada. O resultado esperado é `True`, indicando que nenhuma ocorrência foi perdida durante o agrupamento e a construção da grade.


# Passo 3: Integração Populacional e Criação da Variável-Alvo.

```python
print("--- Passo 3: Integração Populacional e Criação da Variável-Alvo ---")

# 1. Realizar o Merge (Left Join) da grade com a base de população
# Usamos 'municipio' e 'ano' como chaves para garantir que a população mude a cada ano
df_final = df_agrupado.merge(df_populacao[['municipio', 'ano', 'populacao']],
                             on=['municipio', 'ano'],
                             how='left')

# 2. Calcular a Taxa Mensal por 100 mil habitantes (Variável-alvo principal)
# Fórmula da metodologia: Taxa100k = (Ocorrências mensais / População anual) * 100.000
df_final['taxa_100k_total'] = (df_final['total_ocorrencias'] / df_final['populacao']) * 100000

# 3. Calcular a Taxa Mensal apenas para crimes no Ambiente Web (Análise complementar)
df_final['taxa_100k_web'] = (df_final['ocorrencias_web'] / df_final['populacao']) * 100000

# 4. Arredondar as taxas para 2 casas decimais (melhora a visualização e modelagem)
df_final['taxa_100k_total'] = df_final['taxa_100k_total'].round(2)
df_final['taxa_100k_web'] = df_final['taxa_100k_web'].round(2)

# 5. Validar os resultados
print(f"Dimensão após integração populacional: {df_final.shape}")
print(f"Existem municípios sem população mapeada (NaN)? {df_final['populacao'].isnull().sum()} registros vazios")

print("\nAmostra dos Dados com a Taxa Calculada (Head):")
display(df_final.head())

print("\nEstatísticas Descritivas das Taxas:")
display(df_final[['taxa_100k_total', 'taxa_100k_web']].describe())
```

Nesta etapa, a grade espaço-temporal é integrada às estimativas anuais de população utilizando município e ano como chaves. A validação de valores ausentes permite verificar se todos os municípios e anos foram corretamente associados à população correspondente.

A partir dessa integração são calculadas as taxas mensais de ocorrências por 100 mil habitantes, tanto para o conjunto total de registros quanto para aqueles classificados como relacionados ao ambiente Web. Essas taxas constituem as principais variáveis utilizadas nas análises e na modelagem posterior.


# Passo 4: Engenharia de Características

```python
print("--- Passo 4: Engenharia de Características (Lags e Médias Móveis) ---")

# 1. Garantir que os dados estão perfeitamente ordenados no tempo
df_final = (
    df_final
    .sort_values(
        by=['municipio', 'ano', 'mes']
    )
    .reset_index(drop=True)
)

# 2. Criar as variáveis de Lag (Taxas dos meses anteriores)
# O shift(1) "empurra" o dado do mês anterior para a linha do mês atual
df_final['taxa_lag_1'] = (
    df_final
    .groupby('municipio')['taxa_100k_total']
    .shift(1)
)

df_final['taxa_lag_2'] = (
    df_final
    .groupby('municipio')['taxa_100k_total']
    .shift(2)
)

df_final['taxa_lag_3'] = (
    df_final
    .groupby('municipio')['taxa_100k_total']
    .shift(3)
)

# 3. Criar as Médias Móveis dos últimos 3 e 6 meses
# As médias utilizam apenas meses anteriores ao período atual.
# min_periods=3 e min_periods=6 garantem que a média só seja
# calculada quando existir o histórico completo da respectiva janela.
df_final['media_movel_3m'] = (
    df_final
    .groupby('municipio')['taxa_lag_1']
    .transform(
        lambda x: x.rolling(
            window=3,
            min_periods=3
        ).mean()
    )
)

df_final['media_movel_6m'] = (
    df_final
    .groupby('municipio')['taxa_lag_1']
    .transform(
        lambda x: x.rolling(
            window=6,
            min_periods=6
        ).mean()
    )
)

# 4. Arredondar os novos campos para 2 casas decimais
colunas_novas = [
    'taxa_lag_1',
    'taxa_lag_2',
    'taxa_lag_3',
    'media_movel_3m',
    'media_movel_6m'
]

df_final[colunas_novas] = (
    df_final[colunas_novas]
    .round(2)
)

# 5. Validar os resultados
print(
    f"Colunas atuais na base: "
    f"{df_final.columns.tolist()}"
)

print(
    "Total de valores nulos gerados "
    "(esperado devido aos primeiros meses sem histórico):"
)

print(
    df_final[colunas_novas]
    .isnull()
    .sum()
)

print(
    "\nAmostra dos Dados com as Variáveis Temporais "
    "(Focando em um município para ver a evolução):"
)

display(
    df_final[
        [
            'municipio',
            'ano',
            'mes',
            'taxa_100k_total',
            'taxa_lag_1',
            'taxa_lag_2',
            'taxa_lag_3',
            'media_movel_3m',
            'media_movel_6m'
        ]
    ].head(8)
)
```

Nesta etapa, foram construídas variáveis temporais utilizando exclusivamente informações anteriores ao mês que se deseja estimar.

Os lags de 1, 2 e 3 meses representam diretamente as taxas observadas nos períodos anteriores. Já as médias móveis de 3 e 6 meses são calculadas sobre a taxa defasada em um mês, impedindo que o valor do próprio período-alvo seja utilizado na construção da feature.

Para manter o significado temporal das variáveis, a média móvel de 3 meses só é calculada quando existem três meses anteriores disponíveis, enquanto a média móvel de 6 meses exige seis meses completos de histórico. Por esse motivo, são esperados valores nulos nos primeiros meses de cada município.

Com 78 municípios, os valores nulos esperados são:
- `taxa_lag_1`: 78;
- `taxa_lag_2`: 156;
- `taxa_lag_3`: 234;
- `media_movel_3m`: 234;
- `media_movel_6m`: 468.

Esses valores são consequência natural da construção das séries temporais e não representam dados faltantes na base original. As linhas sem histórico suficiente serão excluídas apenas na etapa de modelagem.

# Passo 6: Modelagem

```python
print("--- Passo 6: Modelagem (Treino, Teste e Avaliação) ---")
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# 1. Tratar valores nulos gerados pelos Lags (não podemos treinar com NaN)
df_model = df_final.dropna(subset=['taxa_lag_3', 'media_movel_6m']).copy()

# 2. Transformar a variável 'municipio' em colunas numéricas (One-Hot Encoding)
# Isso permite que os modelos interpretem a dimensão espacial (qual cidade é qual)
df_model = pd.get_dummies(df_model, columns=['municipio'], drop_first=True)

# 3. Divisão Temporal Rígida (Sem vazamento de dados)
df_train = df_model[df_model['ano'] < 2025].copy() # 2023 e 2024
df_test = df_model[df_model['ano'] == 2025].copy()  # 2025

# Definir as colunas de entrada (Features - X) e a variável-alvo (Target - y)
colunas_excluir = ['ano', 'total_ocorrencias', 'ocorrencias_web', 'taxa_100k_total', 'taxa_100k_web']
features = [col for col in df_model.columns if col not in colunas_excluir]

X_train = df_train[features]
y_train = df_train['taxa_100k_total']
X_test = df_test[features]
y_test = df_test['taxa_100k_total']

# 4. Construir o Baseline: "Amanhã será igual a hoje" (taxa do mês anterior)
y_pred_baseline = df_test['taxa_lag_1']

# 5. Treinar os Algoritmos de Machine Learning
# Random Forest
print("Treinando Random Forest...")
rf_model = RandomForestRegressor(random_state=42, n_estimators=100)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)

# XGBoost
print("Treinando XGBoost...")
xgb_model = XGBRegressor(random_state=42, n_estimators=100)
xgb_model.fit(X_train, y_train)
y_pred_xgb = xgb_model.predict(X_test)

# 6. Avaliar os Resultados
def avaliar_modelo(nome, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f"{nome:<25} -> MAE: {mae:.2f} | RMSE: {rmse:.2f} | R2: {r2:.4f}")

print("\n--- Desempenho no Período de Teste (2025) ---")
avaliar_modelo("Baseline (Mês Anterior)", y_test, y_pred_baseline)
avaliar_modelo("Random Forest", y_test, y_pred_rf)
avaliar_modelo("XGBoost", y_test, y_pred_xgb)
```

Analisando as métricas obtidas no teste com os dados de 2025:

Ganho Preditivo Comprovado: O Baseline (que assume que o mês atual será igual ao mês anterior) obteve o maior erro médio absoluto (MAE: 22.71) e explicou apenas 28% da variância das taxas ($R^2$: 0.2826). Isso comprova a premissa da sua metodologia de que métodos baseados exclusivamente no histórico imediato são insuficientes.  

O Vencedor (Random Forest): O Random Forest teve o melhor desempenho geral, reduzindo o erro médio (MAE) para 19.97 e o RMSE para 26.42.
